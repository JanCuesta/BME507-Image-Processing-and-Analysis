```python
# ------------------------------
# EXERCISE 1 - IMAGE INPUT / OUTPUT
# ------------------------------

!pip install pydicom -q

import os
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd

import pydicom

from PIL import Image
```

```python
# ------------------------------
# UNZIP DATASET
# ------------------------------

!unzip -q "/content/chestimage.zip" -d "/content/"

print(os.listdir("/content/chestimage"))
```

```python
# ------------------------------
# DATASET PATHS
# ------------------------------

png_path = "/content/chestimage/chestimage_PNG"
jpg_path = "/content/chestimage/chestimage_JPG"
dicom_path = "/content/chestimage/chestimages_DICOM"

png_files = sorted([
    os.path.join(png_path, file)
    for file in os.listdir(png_path)
])

jpg_files = sorted([
    os.path.join(jpg_path, file)
    for file in os.listdir(jpg_path)
])

dicom_files = sorted([
    os.path.join(dicom_path, file)
    for file in os.listdir(dicom_path)
])

print("PNG images:", len(png_files))
print("JPG images:", len(jpg_files))
print("DICOM images:", len(dicom_files))
```

```python
# ------------------------------
# READ PNG IMAGE
# ------------------------------

png_image = Image.open(png_files[0])

png_array = np.array(png_image)

print("PNG")
print("Shape:", png_array.shape)
print("Data type:", png_array.dtype)
print("Minimum value:", png_array.min())
print("Maximum value:", png_array.max())
print("Image mode:", png_image.mode)
```

```python
# ------------------------------
# READ JPG IMAGE
# ------------------------------

jpg_image = Image.open(jpg_files[0])

jpg_array = np.array(jpg_image)

print("JPG")
print("Shape:", jpg_array.shape)
print("Data type:", jpg_array.dtype)
print("Minimum value:", jpg_array.min())
print("Maximum value:", jpg_array.max())
print("Image mode:", jpg_image.mode)
```

```python
# ------------------------------
# READ DICOM IMAGE
# ------------------------------

dicom = pydicom.dcmread(dicom_files[0])

dicom_array = dicom.pixel_array

print("DICOM")
print("Shape:", dicom_array.shape)
print("Data type:", dicom_array.dtype)
print("Minimum value:", dicom_array.min())
print("Maximum value:", dicom_array.max())

print("\nDICOM information:")
print("Rows:", dicom.Rows)
print("Columns:", dicom.Columns)
print("Bits stored:", dicom.BitsStored)
print("Photometric interpretation:", dicom.PhotometricInterpretation)
```

```python
# ------------------------------
# DISPLAY PNG, JPG AND DICOM
# ------------------------------

plt.figure(figsize=(6, 6))
plt.imshow(png_array, cmap='gray')
plt.title('PNG Image')
plt.axis('off')
plt.show()


plt.figure(figsize=(6, 6))
plt.imshow(jpg_array, cmap='gray')
plt.title('JPG Image')
plt.axis('off')
plt.show()


plt.figure(figsize=(6, 6))
plt.imshow(dicom_array, cmap='gray')
plt.title('DICOM Image')
plt.axis('off')
plt.show()
```

```python
# ------------------------------
# COMPARE IMAGE PROPERTIES
# ------------------------------

comparison = pd.DataFrame({
    'Format': ['PNG', 'JPG', 'DICOM'],

    'Shape': [
        str(png_array.shape),
        str(jpg_array.shape),
        str(dicom_array.shape)
    ],

    'Data Type': [
        str(png_array.dtype),
        str(jpg_array.dtype),
        str(dicom_array.dtype)
    ],

    'Minimum Pixel Value': [
        png_array.min(),
        jpg_array.min(),
        dicom_array.min()
    ],

    'Maximum Pixel Value': [
        png_array.max(),
        jpg_array.max(),
        dicom_array.max()
    ],

    'File Size (KB)': [
        os.path.getsize(png_files[0]) / 1024,
        os.path.getsize(jpg_files[0]) / 1024,
        os.path.getsize(dicom_files[0]) / 1024
    ]
})

comparison
```

```python
# ------------------------------
# WRITE PNG IMAGE
# ------------------------------

png_image.save("/content/output_png.png")

print("PNG saved.")
```

```python
# ------------------------------
# WRITE JPG IMAGE
# ------------------------------

jpg_image.save(
    "/content/output_jpg.jpg",
    quality=95
)

print("JPG saved.")
```

```python
# ------------------------------
# WRITE DICOM IMAGE
# ------------------------------

dicom.save_as("/content/output_dicom.dcm")

print("DICOM saved.")
```

```python
# ------------------------------
# VERIFY WRITTEN FILES
# ------------------------------

output_png = Image.open("/content/output_png.png")
output_jpg = Image.open("/content/output_jpg.jpg")
output_dicom = pydicom.dcmread("/content/output_dicom.dcm")

print("Output PNG:", np.array(output_png).shape)

print("Output JPG:", np.array(output_jpg).shape)

print(
    "Output DICOM:",
    output_dicom.pixel_array.shape
)
```

```python
# ------------------------------
# COMPARE PNG AND JPG PIXEL VALUES
# ------------------------------

png_compare = np.array(output_png).astype(np.int16)
jpg_compare = np.array(output_jpg).astype(np.int16)

difference = np.abs(
    png_compare - jpg_compare
)

print("Mean absolute pixel difference:", difference.mean())
print("Maximum pixel difference:", difference.max())
print(
    "Percentage of different pixels:",
    np.mean(difference > 0) * 100,
    "%"
)
```
